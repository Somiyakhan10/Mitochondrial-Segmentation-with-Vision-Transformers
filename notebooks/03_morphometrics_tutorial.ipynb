# 03 — Morphometrics Tutorial

Covers FR-19–FR-24, NFR-07: extracting per-mitochondrion shape features from
a segmentation mask, computing dysfunction indices, and flagging
out-of-range measurements. All real and runnable today on any binary mask —
this example uses a synthetic mask since no trained segmentation model
exists yet.

```python
import numpy as np
from mitomorph.morphometrics.single_features import extract_single_features
from mitomorph.morphometrics.dysfunction_indices import fragmentation_index, mitochondrial_density, swelling_score
from mitomorph.morphometrics.quality_control import validate_features

synthetic_mask = np.zeros((100, 100), dtype=bool)
synthetic_mask[10:20, 10:20] = True
synthetic_mask[50:60, 50:70] = True

features = extract_single_features(synthetic_mask)
for f in features:
    print(f)
```

## Dysfunction indices (FR-22–FR-24)

```python
total_area = float(synthetic_mask.size)
print("Fragmentation index:", fragmentation_index(features, total_area))
print("Mitochondrial density:", mitochondrial_density(features, total_area))
```

## Quality control (NFR-07)

```python
qc_results = validate_features(features)
for r in qc_results:
    print(r)
```
